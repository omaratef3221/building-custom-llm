## Masked Multi Head Attention Testing

```python
from Layers.MultiheadAttention import MultiHeadAttention

d_model = 4
attention_layer = MultiHeadAttention(d_model, 2)
test_input = torch.rand((2, 3, d_model))

mask = torch.tril(torch.full((3, 3), float('-inf')), diagonal=1)
print(attention_layer(test_input, test_input, test_input, mask = mask).shape)
print(attention_layer(test_input,test_input, test_input, mask = mask))
```
